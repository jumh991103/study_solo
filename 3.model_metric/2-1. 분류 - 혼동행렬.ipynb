# 1. 모델 성능 평가지표(Metric)
> 실제값과 모델에 의한 예측값을 비교하여 모델의 성능을 측정하는 방법

# 2. 분류 모델 성능 평가 지표 

## 2-1. 혼동행렬(Confusion Matrix)
> 분석 모델에서 구한 분류의 예측 범주와 데이터의 실제 분류 범주를 교차 표 형태로 정리한 행렬  
> 범주형의 경우 숫자가 아니라 오차를 못 구함(연산이 안됨). 그래서 해당표로 학습 끝나고 나면 평가함.

- Actual : 실제 데이터
- Predicted : 예측 데이터
- Negative vs Positive : target데이터에서 우리가 찾고자하는 값을 positive(1)로 둠.  
    > 예를 들자면 타이타닉에서 살아남은 자
- 혼동행렬 해석할땐 항상 predicted가 주어임.  
    > 예를 들어, 모델이 positive라고 예측한 데이터가 실제로 positive인 경우(모델이 맞춘 경우)엔 'True Positive.'  
- Accuracy : 정확도 
    > TP+TN(맞춘거) / TP+TN+FP+FN(전체)
- SPecificity : 특이도
    > TN
- Recall, Sensitivity : 민감도
- Precision :정밀도
