>학습이 끝난 뒤(9/3에 배운 것): 같은 MSE를 평가지표로 씀

# 1. 선형 회귀
- 단순 선형 회귀식  
$$
\hat{y} = wx + b
$$

> 선형회귀에서는 MSE를 최소화하는 방법으로 w와 b를 구할 수 있음  
> MSE식 중에서 (예측값 - 실제값)^2 부분이 손실함수를 의미한다.  
> 모델이 학습한다는 뜻은 손실함수의 값을 작게 변하도록 학습을 진행하는 거라고 보면됨.  
  

$w$의 최소값을 구하는 방법은?
1. MSE = ${1 \over N}\sum(\hat{y_i} - y_i)^2$
2. => ${1 \over N}\sum(wx + b - y_i)^2$
3. 미분(w) => $\frac{d}{dw}({1 \over N}\sum(wx + b - y_i)^2)$
4. $w$의 gradient => $2{1 \over N}\sum(wx + b - y_i)x$
5. learning_rate: w의 gradient를 업데이트할때, 얼마나 업데이트를 할지 정의하는 변수
6. 실제로 업데이트할 $w$의 gradient는
> - learning_rate * $2{1 \over N}\sum(\hat{y_i} - y_i)x$입니다.
> - $2{1 \over N}\sum$는 평균으로 해석할 수 있으므로,
> - Learning Rate * 평균((Y예측 - Y실제) * X)로 해석할 수 있다.  

 - Learning Rate(step size) - 몇 퍼센트
 >러닝 레이트가 너무 작으면, 학습이 오래걸리고  
 >러닝 레이트가 너무 크면, 학습이 되지않고 끝없이 진동할 수 있음.  
   
 >실제 모델에선 오차를 바로 안적용하고 learning rate를 설정하여 비율을 적용함. 즉 0.7 learning rate는   
 >오차의 70프로만 모델에 적용하겠다는 말임.
